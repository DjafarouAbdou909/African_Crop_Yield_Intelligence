Après la compréhension du métier, nous allons passer à la compréhension des données.

```python
import pandas as pd
from IPython.display import display

pd.set_option("display.max_columns", 30)
```

```python
DATA_DIRS = "../data/raw/"
FILES = {
    "cropcut" : "GROW-Africa_LSMS_cropcut.xlsx",
    "survey" : "GROW-Africa_LSMS_survey.xlsx",
    "point" : "GROW-Africa_Point.xlsx",
    "regional" : "GROW-Africa_Regional.xlsx"
}
```

```python
data = {}
for name, filename in FILES.items():
    data[name] = pd.read_excel(DATA_DIRS + filename)
    print(f"Chargement réussi : {name}")
```

```python
for name, current_df in data.items():
    print(f" STRUCTURE ET APERÇU DU FICHIER : {name.upper()} ")
    
    # Infos structurelles
    print("\n -----INFO STRUCTURE--------")
    current_df.info()
    
    # Aperçu des premières lignes
    print("\n --------APERÇU DES DONNÉES---------")
    display(current_df.head(3))  
    
    # Statistiques descriptives
    print("\n --------------STATISTIQUES DESCRIPTIVES----------")
    display(current_df.describe(include='all'))  
```

**Observation sur info():**
- Les quatre fichiers ont bien le nombre de lignes attendu : cropcut (11 345), survey (195 152), point (131 812) et regional (197 535).
- La cible de rendement est complète dans `cropcut` (100 %) et `point` (100 %). Elle est manquante pour environ 19 % des lignes de `survey` (`Yield_ton_ha_`) et 17 % de `regional` (`yieldTonHa`).
- Les coordonnées GPS sont incomplètes : environ 41 % de manquants dans `point` (`Latitude`, `Longitude`), environ 13 % dans `survey`, moins de 1 % dans `cropcut`.
- Plusieurs colonnes sont presque vides : `L3_GID` dans `regional` (98 % de manquants), `ProdEst2_kg_` et `YieldEst2UsingGPS_ton_ha_` dans `survey` (plus de 94 % de manquants).
- Les identifiants `holder_id` et `household_id` de `cropcut` sont lus en `float64` (25 % de manquants) et les dates de `point` (`PlantDate`, `HarvestDate`) sont lues en nombres.

**Décision :**
- Conserver les lignes sans rendement pour l'instant, les exclure seulement au moment de la modélisation.
- Ne pas imputer les coordonnées GPS ici, vérifier d'abord si le niveau administratif (`L1_GID`, `L2_GID`) suffit.
- Écarter les colonnes presque vides des variables explicatives.
- Corriger les types (identifiants en texte, dates) dans la phase de préparation des données.

**Observation sur describe():**

*cropcut*
- Le rendement est plausible : moyenne 1,13 t/ha, médiane 0,86, maximum 10,75.
- `cropCutMonth` va jusqu'à 83 et commence à 0 : un mois doit être compris entre 1 et 12, il y a des valeurs invalides.
- `freshWeightKg` atteint 200 kg alors que `dryWeightKg` ne dépasse pas 17,2 kg : valeur extrême à vérifier.
- `household_id` est égal à `holder_id` divisé par 100 (même moyenne, même structure) : les deux colonnes sont redondantes.

*survey*
- Valeurs impossibles : `prcCrop` va jusqu'à 1000 (un pourcentage ne dépasse pas 100), `ReportedArea_ha_` jusqu'à 5,99 milliards d'hectares, `Yield_ton_ha_` jusqu'à 155,7 t/ha.
- Distributions très asymétriques : `Yield_ton_ha_` a une moyenne de 1,50 pour une médiane de 0,74, et `ReportedArea_ha_` une moyenne de 35 489 pour une médiane de 0,40.

*point*
- `PlantDate` et `HarvestDate` sont des numéros de série Excel (environ 43 000, soit 2018), avec des valeurs aberrantes : minimum 272 (année 1900) et maximum 73 170 (année 2200).
- `Longitude` va jusqu'à 76,1 : ce point est hors d'Afrique, les coordonnées sont erronées.
- `HarvestYear` commence en 1974, alors que `cropcut` couvre 2012-2019.
- `Yield_ton_ha_` : moyenne 2,55, médiane 1,73, maximum 150 t/ha.

*regional*
- Les valeurs -999 et -99 servent de codes de données manquantes (minimum de `areaHa`, `productionTon` et `yieldTonHa`) : ce ne sont pas de vraies mesures.
- `yieldTonHa` atteint 6 996 t/ha (moyenne 3,26, médiane 1,0) et `areaHa` 1,75 milliard d'hectares : valeurs impossibles.
- `harvest_year` couvre 1960-2023 : période beaucoup plus large que les autres fichiers.

**Décision :**
- Remplacer -999 et -99 par des valeurs manquantes dans `regional`.
- Convertir `PlantDate` et `HarvestDate` en vraies dates et invalider celles qui sont hors de la plage plausible.
- Marquer comme invalides `cropCutMonth` hors de 1-12, `prcCrop` au-dessus de 100 et les coordonnées hors d'Afrique.
- Fixer un seuil de rendement plausible par culture avant de traiter les valeurs extrêmes
- Retirer `SpatialPrecision_km_` de `survey` et `household_id` de `cropcut`.
- Ne pas modifier les données brutes ici : toutes ces corrections se font dans la phase de préparation.
