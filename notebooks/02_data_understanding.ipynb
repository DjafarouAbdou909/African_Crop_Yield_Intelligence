```python
cles_candidates = {
    "cropcut": [
        ["holder_id", "household_id", "parcel_id", "field_id"]
    ],
    "survey": [
        ["hhid", "parcel_id"],
        ["hhid", "plotid"]
    ]
}

for nom, cles in cles_candidates.items():
    df = data[nom]
    print(f"\n--- {nom.upper()} ---")

    for colonnes in cles:
        # On évalue uniquement les lignes où toutes les clés sont renseignées.
        df_complet = df.dropna(subset=colonnes)

        print(f"\nClé candidate : {colonnes}")
        print("Lignes avec clé complète :", len(df_complet))
        print(
            "Doublons sur cette combinaison :",
            df_complet.duplicated(subset=colonnes).sum()
        )
```

#### 3.2 Construction des rendements dans `survey`

```python
def part_conforme(estimation, reference, tolerance=1e-3):
    # Part des lignes où l'estimation reproduit la référence à la tolérance relative près.
    comparables = estimation.notna() & reference.notna() & np.isfinite(estimation)
    ecart = (estimation[comparables] - reference[comparables]).abs() / reference[comparables].abs()
    return comparables.sum(), (ecart < tolerance).mean()


part_culture = survey["prcCrop"] / 100
formules = {
    "Yield = Prod / ReportedArea": (survey["Prod_kg_"] / survey["ReportedArea_ha_"] / 1000, "Yield_ton_ha_"),
    "Yield = Prod / (ReportedArea × prcCrop)": (survey["Prod_kg_"] / (survey["ReportedArea_ha_"] * part_culture) / 1000, "Yield_ton_ha_"),
    "YieldUsingGPS = Prod / (GPSArea × prcCrop)": (survey["Prod_kg_"] / (survey["GPSArea_ha_"] * part_culture) / 1000, "YieldUsingGPS_ton_ha_"),
    "YieldEst2 = ProdEst2 / (ReportedArea × prcCrop)": (survey["ProdEst2_kg_"] / (survey["ReportedArea_ha_"] * part_culture) / 1000, "YieldEst2_ton_ha_"),
    "YieldEst2UsingGPS = ProdEst2 / (GPSArea × prcCrop)": (survey["ProdEst2_kg_"] / (survey["GPSArea_ha_"] * part_culture) / 1000, "YieldEst2UsingGPS_ton_ha_"),
}
for nom, (estimation, cible) in formules.items():
    n, part = part_conforme(estimation, survey[cible])
    print(f"{nom:<52} n = {n:>7} | conformes : {part:.2%}")

estimation = survey["Prod_kg_"] / (survey["ReportedArea_ha_"] * part_culture) / 1000
ecart_relatif = (estimation - survey["Yield_ton_ha_"]).abs() / survey["Yield_ton_ha_"]
non_conformes = survey[np.isfinite(estimation) & (ecart_relatif > 1e-3)]
print("\nNon conformes à la formule principale :", len(non_conformes))
print("Principales sources :", non_conformes["Source"].value_counts().head(4).to_dict())
print("Rapport Yield / estimation (quartiles) :",
      (non_conformes["Yield_ton_ha_"] / estimation[non_conformes.index]).quantile([0.25, 0.5, 0.75]).round(2).to_dict())

sans_prod2 = survey["YieldEst2_ton_ha_"].notna() & survey["ProdEst2_kg_"].isna()
print("\nYieldEst2_ton_ha_ renseigné sans ProdEst2_kg_ :", sans_prod2.sum(), survey.loc[sans_prod2, "Source"].value_counts().to_dict())
print("prcCrop > 100 :", (survey["prcCrop"] > 100).sum(), survey.loc[survey["prcCrop"] > 100, "Source"].value_counts().to_dict())
```

**Constat**

- `Yield_ton_ha_` est égal au rapport entre `Prod_kg_` et le produit de `ReportedArea_ha_` par la part de culture `prcCrop`, converti en tonnes, pour 99,10 % des lignes comparables. Sans `prcCrop`, la formule ne reproduit que 52,70 % des lignes, essentiellement celles où `prcCrop` vaut 100.
- La même construction reproduit `YieldUsingGPS_ton_ha_` avec la surface GPS pour 99,37 % des lignes, `YieldEst2_ton_ha_` avec `ProdEst2_kg_` pour 99,72 % et `YieldEst2UsingGPS_ton_ha_` pour 99,66 %.
- Les 1 424 lignes non conformes se concentrent dans les vagues `CIV2021`, `MLI2021`, `GNB2021` et `MLI2018`. Le rapport entre rendement publié et rendement recalculé y prend souvent une valeur simple, de médiane égale à 2.
- `YieldEst2_ton_ha_` est renseigné sans `ProdEst2_kg_` sur 24 066 lignes, toutes tanzaniennes. Sa définition n'est donc pas homogène.
- `prcCrop` dépasse 100 sur 80 lignes, toutes issues de `LSMS_MWI2019`.

**Analyse**

- `prcCrop` mesure la part de la parcelle occupée par la culture en situation de cultures associées, ce qui confirme la définition déduite dans le dictionnaire des données.
- Les rapports simples observés sur les lignes non conformes évoquent des conversions d'unités locales. Cette hypothèse reste à confirmer dans la documentation LSMS.

**Conséquences pour la suite**

`Prod_kg_`, `ProdEst2_kg_`, `ReportedArea_ha_`, `GPSArea_ha_` et `prcCrop` entrent dans le calcul des cibles et constituent des fuites certaines. Les quatre colonnes de rendement sont des variantes d'une même mesure. Une seule sera retenue comme cible et les autres seront exclues des prédicteurs.

```python
# Bornes du filtre décrit en page 4 de l'article, reconstruites à partir du fichier regional.
# Un rendement est aberrant s'il est inférieur à 10 % de la médiane ou supérieur à 200 % du 98e percentile
# de la distribution régionale de la culture.
rendements_regionaux = regional["yieldTonHa"].where(regional["yieldTonHa"] > 0)
bornes = rendements_regionaux.groupby(regional["crop"]).agg(
    mediane="median",
    p98=lambda s: s.quantile(0.98),
)
bornes["borne_basse"] = 0.10 * bornes["mediane"]
bornes["borne_haute"] = 2.00 * bornes["p98"]

seuils = survey[["Crop"]].join(bornes, on="Crop")
hors_bornes_calcule = (estimation < seuils["borne_basse"]) | (estimation > seuils["borne_haute"])
hors_bornes_publie = (survey["Yield_ton_ha_"] < seuils["borne_basse"]) | (survey["Yield_ton_ha_"] > seuils["borne_haute"])

cible_absente_calculable = survey["Yield_ton_ha_"].isna() & np.isfinite(estimation)
print("Rendement absent mais calculable à partir de Prod, ReportedArea et prcCrop :", cible_absente_calculable.sum())
print(f"  dont hors des bornes du filtre : {hors_bornes_calcule[cible_absente_calculable].mean():.2%}")
print(f"Rendement présent mais hors des bornes : {hors_bornes_publie[survey['Yield_ton_ha_'].notna()].mean():.2%}")

for nom, df in [("point", point), ("cropcut", cropcut)]:
    seuils_df = df[["Crop"]].join(bornes, on="Crop")
    hors = (df["Yield_ton_ha_"] < seuils_df["borne_basse"]) | (df["Yield_ton_ha_"] > seuils_df["borne_haute"])
    print(f"{nom} : rendements hors bornes = {hors.mean():.2%}")
```

**Constat**

Parmi les 10 550 lignes de `survey` dont le rendement est absent mais calculable, 96,48 % ont un rendement recalculé situé hors des bornes du filtre décrit par les auteurs. À l'inverse, seulement 0,73 % des rendements publiés de `survey` se situent hors de ces bornes. Ce taux est de 0,24 % dans `point` et de 0,02 % dans `cropcut`.

**Analyse**

L'absence de `Yield_ton_ha_` dans `survey` résulte principalement du filtrage des valeurs aberrantes opéré par les auteurs. Ces valeurs ne manquent donc pas au hasard. Elles manquent en raison de leur caractère extrême, ce qui correspond à un mécanisme MNAR. Le niveau de preuve est élevé, mais les bornes sont reconstruites faute de seuils publiés et le traitement des codes négatifs par les auteurs n'est pas documenté. La correspondance n'est donc pas parfaite.

**Conséquences pour la suite**

- La cible manquante de `survey` ne sera ni imputée ni recalculée à partir de la production et des surfaces, ce qui réintroduirait des valeurs jugées aberrantes par les auteurs.
- La distribution publiée est tronquée. Un modèle entraîné sur ces données sous-estimera l'incertitude dans les valeurs extrêmes, ce que l'analyse des erreurs devra signaler.
- La décision initiale de conserver les lignes sans rendement reste valable pour l'audit. Ces lignes ne pourront toutefois pas servir à la modélisation, leur seule exploitation possible passant par des variables exclues pour fuite.

#### 3.3 Codes spéciaux, cohérence entre production, surface et rendement et indicateurs qualité dans `regional`

```python
for colonne in ["areaHa", "productionTon", "yieldTonHa"]:
    valeurs = regional[colonne]
    autres_negatifs = ((valeurs < 0) & ~valeurs.isin([-999, -99])).sum()
    print(f"{colonne:<14} -999 : {(valeurs == -999).sum():>4} | -99 : {(valeurs == -99).sum():>3} | "
          f"autres négatifs : {autres_negatifs:>3} | zéros : {(valeurs == 0).sum()}")

negatifs = regional[(regional["yieldTonHa"] < 0) & (regional["yieldTonHa"] != -99)]
explique = (
    np.isclose(negatifs["yieldTonHa"], negatifs["productionTon"] / negatifs["areaHa"])
    | np.isclose(negatifs["yieldTonHa"], negatifs["areaHa"] / negatifs["productionTon"])
)
contient_999 = negatifs[["areaHa", "productionTon"]].eq(-999).any(axis=1)
print("\nRendements négatifs autres que -99 :", len(negatifs))
print("  dont areaHa ou productionTon == -999 :", contient_999.sum())
print("  dont égal à P/A ou A/P :", explique.sum())
print("Sources des codes -999 :", regional.loc[regional[["areaHa", "productionTon"]].eq(-999).any(axis=1), "Source"].value_counts().to_dict())
print("Sources des codes -99  :", regional.loc[regional[["productionTon", "yieldTonHa"]].eq(-99).any(axis=1), "Source"].value_counts().to_dict())
```

**Constat**

- Le code -999 n'apparaît que dans `AgroMAPS`, sur la surface et la production. Le code -99 n'apparaît que dans `RCeA`, sur la production et le rendement.
- Les 30 rendements négatifs distincts de -99, tels que -0,01001, résultent tous d'une division dont l'un des termes vaut -999. À titre d'exemple, -10 divisé par 999 donne -0,01001.

**Analyse**

-999 et -99 sont des codes de valeur manquante, ce qui confirme la décision de l'audit initial. Ils ont en outre été intégrés au calcul du rendement et ont produit des valeurs négatives dépourvues de sens.

**Conséquences pour la suite**

La préparation remplacera -999 et -99 par des valeurs manquantes dans les trois colonnes et invalidera tout rendement négatif dérivé de ces codes. Le seul remplacement des codes visibles serait insuffisant.

```python
testables = (regional["areaHa"] > 0) & (regional["productionTon"] > 0) & (regional["yieldTonHa"] > 0)
production_sur_surface = regional["productionTon"] / regional["areaHa"]
surface_sur_production = regional["areaHa"] / regional["productionTon"]

regional_diag = regional.assign(
    relation=np.select(
        [
            ~testables,
            np.isclose(regional["yieldTonHa"], production_sur_surface, rtol=1e-3),
            np.isclose(regional["yieldTonHa"], surface_sur_production, rtol=1e-3),
        ],
        ["non testable", "Y = P / A", "Y = A / P"],
        default="autre",
    ),
    production_sur_surface=production_sur_surface,
)
display(pd.crosstab(regional_diag["Source"], regional_diag["relation"], margins=True, margins_name="Total"))
```

```python
# Si Y = A / P, soit le rendement est faux, soit areaHa et productionTon sont inversées.
# Les tubercules, à fort rendement, permettent de trancher car un manioc à 0,1 t/ha n'est pas plausible.
cultures_test = ["Cassava", "Yam", "Potato", "Sweet potato", "Maize", "Sorghum", "Rice", "Beans"]
comparaison = (
    regional_diag[
        regional_diag["relation"].isin(["Y = P / A", "Y = A / P"])
        & regional_diag["crop"].isin(cultures_test)
    ]
    .groupby(["crop", "relation"])[["yieldTonHa", "production_sur_surface"]]
    .median()
    .unstack("relation")
    .round(3)
)
display(comparaison)
```

**Constat**

- `yieldTonHa` est égal au rapport entre `productionTon` et `areaHa` pour 126 696 lignes. Pour 33 740 lignes, soit 17,1 % du fichier, il est en revanche égal au rapport inverse entre `areaHa` et `productionTon`.
- Cette inversion concerne `HarvestStat` pour 25 631 lignes, `RCeA` pour 3 262, `FAO_subnational` pour 3 233, soit 44,5 % de cette source, `AgroMAPS` pour 1 579 et `USAID_KadunaState` pour 35. `FAOSTAT` n'est pas concernée.
- Dans le groupe inversé, le rendement médian du manioc atteint 9,98 t/ha, du même ordre que dans le groupe conforme où il s'établit à 8,41 t/ha. Le rapport entre production et surface y serait de 0,10 t/ha. Le même contraste s'observe pour l'igname, la pomme de terre et la patate douce.
- 1 710 lignes ne vérifient aucune des deux relations, dont 1 623 issues de `RCeA` et 87 de `FAOSTAT`.

**Analyse**

Pour les lignes inversées, le rendement est plausible et ce sont les colonnes `areaHa` et `productionTon` qui sont permutées. Un rendement de manioc de 0,1 t/ha est agronomiquement impossible. Le niveau de preuve est élevé pour les tubercules. Pour les céréales, dont le rendement avoisine 1 t/ha, les deux lectures restent individuellement possibles.

**Conséquences pour la suite**

- `yieldTonHa` peut servir de cible sans recalcul.

```python
motif_gid = r"L[0-3][A-Z]{3}\d+"
for nom, df in data.items():
    print(f"\n--- {nom.upper()} ---")
    for colonne in ["L1_GID", "L2_GID", "L3_GID"]:
        valeurs = df[colonne].dropna()
        bien_formes = valeurs.str.fullmatch(motif_gid)
        mal_formes = valeurs[~bien_formes]
        pays_different = bien_formes & (valeurs.str[2:5] != df.loc[valeurs.index, "L0_GID"].str[2:5])
        print(f"{colonne} : mal formés = {len(mal_formes):>3} {mal_formes.map(repr).value_counts().head(2).to_dict()} "
              f"| code pays différent de L0_GID = {pays_different.sum()}")
```

```python
colonnes_gid = ["L0_GID", "L1_GID", "L2_GID", "L3_GID"]
for nom, df in data.items():
    gid = df[colonnes_gid].replace({"[]": np.nan})
    for colonne in colonnes_gid:
        gid[colonne] = gid[colonne].str.replace("﻿", "", regex=False)
    violations = {}
    for enfant, parent in [("L1_GID", "L0_GID"), ("L2_GID", "L1_GID"), ("L3_GID", "L2_GID")]:
        nb_parents = gid.dropna(subset=[enfant, parent]).groupby(enfant)[parent].nunique()
        violations[f"{enfant}->{parent}"] = int((nb_parents > 1).sum())
    print(f"{nom:<9} identifiants rattachés à plusieurs parents (après nettoyage de '[]' et du BOM) : {violations}")
```

#### 6.2 Plausibilité des coordonnées

```python
def diagnostic_coordonnees(df, col_lat, col_lon):
    # Compare chaque point à la médiane des coordonnées de son pays selon L0_GID.
    # Signatures recherchées, latitude et longitude inversées ou signe de la latitude inversé.
    coords = df.dropna(subset=[col_lat, col_lon])
    centre = coords.groupby("L0_GID")[[col_lat, col_lon]].transform("median")
    distance = np.hypot(coords[col_lat] - centre[col_lat], coords[col_lon] - centre[col_lon])
    distance_inversee = np.hypot(coords[col_lon] - centre[col_lat], coords[col_lat] - centre[col_lon])
    distance_signe = np.hypot(-coords[col_lat] - centre[col_lat], coords[col_lon] - centre[col_lon])
    eloigne = distance > 5
    return coords.assign(diagnostic=np.select(
        [eloigne & (distance_inversee < 2), eloigne & (distance_signe < 2)],
        ["latitude/longitude inversées", "signe de latitude inversé"],
        default="pas de signature",
    ))


geolocalises = [
    ("cropcut", cropcut, "GPS_lat", "GPS_lon"),
    ("survey", survey, "GPS_lat", "GPS_lon"),
    ("point", point, "Latitude", "Longitude"),
]
for nom, df, col_lat, col_lon in geolocalises:
    diag = diagnostic_coordonnees(df, col_lat, col_lon)
    hors_afrique = ~diag[col_lat].between(-35, 38) | ~diag[col_lon].between(-26, 52)
    anomalies = diag[diag["diagnostic"] != "pas de signature"]
    print(f"{nom:<8} hors du cadre Afrique : {hors_afrique.sum()} | signatures : "
          f"{anomalies.groupby(['diagnostic', 'Source', 'Country']).size().to_dict()}")
```

**Constat**

- Dans `point`, 215 coordonnées de `Niche_trial`, toutes relatives au site rwandais de Musasa, présentent une latitude et une longitude inversées, avec une latitude de 29,8 et une longitude de -1,5 au lieu de -1,5 et 29,8. 30 coordonnées de `Corbeels2020` au Malawi et en Tanzanie présentent une latitude de signe erroné, Lilongwe étant par exemple placée à 14,2 au lieu de -14,2. Un point `OneAcre_survey` situé au Kenya affiche une longitude de 76,1, hors d'Afrique.
- Aucune de ces signatures n'apparaît dans `cropcut` ni dans `survey`.

**Analyse**

Ces erreurs sont systématiques à l'échelle d'un site ou d'une étude, ce qui est caractéristique d'erreurs de saisie dans une compilation de littérature ou d'essais. Le diagnostic ne détecte que ces deux signatures. Une erreur qui maintient le point dans un pays voisin ou dans le bon pays resterait invisible. Seule une jointure spatiale avec les polygones GADM, non disponibles ici, permettrait de valider chaque point.

**Conséquences pour la suite**

La préparation corrigera les deux signatures identifiées et invalidera le point situé hors d'Afrique. Les variables spatiales s'appuieront de préférence sur les identifiants GID, plus fiables que les coordonnées, d'autant que les coordonnées LSMS sont volontairement déplacées de 0 à 5 km selon l'article.

### 7. Biais de mesure, redondances et représentativité

#### 7.1 Surface déclarée et surface mesurée par GPS dans `survey`

```python
ACRE_HA = 0.40468564224
surface_declaree = survey["ReportedArea_ha_"]
surface_gps = survey["GPSArea_ha_"]


def sur_grille_quart_acre(surface):
    quarts_acre = surface / ACRE_HA * 4
    return np.isclose(quarts_acre, quarts_acre.round(), atol=1e-3) & (surface > 0)


print(f"Surfaces sur une grille de 1/4 d'acre : déclarées {sur_grille_quart_acre(surface_declaree)[surface_declaree > 0].mean():.1%}"
      f" | GPS {sur_grille_quart_acre(surface_gps)[surface_gps > 0].mean():.1%}")
print("Surfaces déclarées les plus fréquentes (ha) :", surface_declaree.round(4).value_counts().head(6).to_dict())

deux_mesures = (surface_declaree > 0) & (surface_gps > 0)
classes_gps = pd.cut(surface_gps[deux_mesures], [0, 0.1, 0.4, 2, np.inf])
synthese_surfaces = pd.DataFrame({
    "ratio_declaree_gps": (surface_declaree / surface_gps)[deux_mesures],
    "rendement_gps": survey.loc[deux_mesures, "YieldUsingGPS_ton_ha_"],
}).groupby(classes_gps, observed=True).agg(
    n=("ratio_declaree_gps", "size"),
    ratio_median=("ratio_declaree_gps", "median"),
    rendement_gps_median=("rendement_gps", "median"),
)
display(synthese_surfaces.round(3))

ligne_max = surface_declaree.idxmax()
print("Surface déclarée maximale :", surface_declaree.max(), "ha | source :", survey.loc[ligne_max, "Source"])
print("Moyenne de ReportedArea_ha_ avec cette ligne :", round(surface_declaree.mean(), 3),
      "| sans cette ligne :", round(surface_declaree.drop(ligne_max).mean(), 3))
```

**Constat**

- 59,8 % des surfaces déclarées correspondent exactement à un multiple d'un quart d'acre, comme 0,1012, 0,2023 ou 0,4047 ha, contre 4,3 % des surfaces mesurées par GPS.
- Le rapport entre surface déclarée et surface GPS dépend de la taille de la parcelle. Sa médiane atteint 1,67 sous 0,1 ha, ce qui traduit une surestimation, et 0,81 au-delà de 2 ha, ce qui traduit une sous-estimation.
- Le rendement médian calculé avec la surface GPS décroît fortement avec la taille de la parcelle, de 1,73 t/ha sous 0,1 ha à 0,34 t/ha au-delà de 2 ha.
- La moyenne de 35 489 ha relevée lors de l'analyse descriptive initiale est imputable à une seule ligne de 5,99 milliards d'hectares issue de `LSMS_GNB2021`. Sans cette ligne, la moyenne s'établit à 0,83 ha.

**Analyse**

- Les surfaces déclarées sont arrondies à des fractions d'acre, ce qui traduit une accumulation sur des valeurs rondes. Ce constat est cohérent avec la figure S19 de l'article, selon laquelle les agriculteurs surestiment les petites parcelles et sous-estiment les grandes.
- La relation inverse entre surface et rendement combine un effet agronomique possible et une erreur de mesure, une petite surface mal mesurée placée au dénominateur gonflant mécaniquement le rendement. Les données ne permettent pas de séparer ces deux effets.

**Conséquences pour la suite**

- Pour `survey`, `YieldUsingGPS_ton_ha_` repose sur une mesure objective de la surface et constitue la cible la plus défendable, au prix d'une couverture réduite à 152 205 lignes contre 158 478.
- La surface n'est pas un prédicteur admissible en raison du risque de fuite. L'erreur sur la cible dépendant de la taille de la parcelle, l'analyse des erreurs sera conduite par classe de surface.

#### 7.2 Comparaison des enquêtes et des statistiques officielles avec les mesures directes en Éthiopie

```python
cle_comparaison = ["L2_GID", "Crop", "annee"]
cc_l2 = (
    cropcut.assign(annee=cropcut["Year"])
    .dropna(subset=["L2_GID"])
    .groupby(cle_comparaison)["Yield_ton_ha_"].agg(["mean", "median", "size"])
)
sv_l2 = (
    survey[survey["Country"] == "Ethiopia"].assign(annee=survey["AgYearEnd"])
    .dropna(subset=["L2_GID", "Yield_ton_ha_"])
    .groupby(cle_comparaison)["Yield_ton_ha_"].agg(["mean", "median", "size"])
)
paires_sv = cc_l2.join(sv_l2, lsuffix="_cc", rsuffix="_sv", how="inner")
paires_sv = paires_sv[(paires_sv["size_cc"] >= 10) & (paires_sv["size_sv"] >= 10)]
ratio_sv = paires_sv["median_sv"] / paires_sv["median_cc"]
print(f"survey / cropcut : {len(paires_sv)} cellules L2 × culture × année | rapport médian des médianes : {ratio_sv.median():.3f} "
      f"| part où survey > cropcut : {(ratio_sv > 1).mean():.1%}")

ethiopie_regional = regional[(regional["country"] == "Ethiopia") & (regional["level"] == 2) & (regional["yieldTonHa"] > 0)]
for decalage in [0, 1]:
    rg_l2 = (
        ethiopie_regional.rename(columns={"crop": "Crop"})
        .assign(annee=ethiopie_regional["harvest_year"] + decalage)
        .groupby(cle_comparaison)["yieldTonHa"].mean().rename("regional")
    )
    paires_rg = cc_l2[cc_l2["size"] >= 10].join(rg_l2, how="inner")
    ratio_rg = paires_rg["regional"] / paires_rg["mean"]
    print(f"regional / cropcut (harvest_year + {decalage}) : {len(paires_rg)} cellules | rapport médian : {ratio_rg.median():.3f} "
          f"| part où regional > cropcut : {(ratio_rg > 1).mean():.1%}")
```

**Constat**

Pour une même zone `L2_GID`, une même culture et une même année, avec au moins 10 observations de part et d'autre, les rendements de `survey` dépassent ceux de `cropcut` dans 69,8 % des cellules, pour un rapport médian de 1,46. Les statistiques officielles de `regional` dépassent `cropcut` dans 88,7 à 90,1 % des cellules, pour un rapport médian compris entre 1,78 et 1,88. Ce résultat varie peu selon la convention d'alignement des années.

**Analyse**

Le sens de l'écart concorde avec l'article, selon lequel enquêtes et statistiques officielles surestiment le rendement par rapport aux mesures directes. Son ampleur est en revanche supérieure aux biais de 20 % et de 32 % publiés. Les méthodes diffèrent, puisque la présente comparaison ne pondère pas les cellules, mélange les saisons et les sources de `regional` et repose sur des rapports de médianes. Ces chiffres ne constituent donc pas une estimation du biais. Ils établissent uniquement que les trois sources ne mesurent pas la même grandeur avec la même erreur.

**Conséquences pour la suite**

- `cropcut`, `survey` et `regional` ne seront pas fusionnés en une cible unique sans correction de biais ou sans variable identifiant le type de source.
- En cas de combinaison des fichiers, le type de source deviendra une variable nécessaire et l'évaluation sera conduite séparément pour chaque type.
- Les mesures directes de `cropcut` demeurent la référence la plus fiable, mais elles ne couvrent qu'un seul pays.

#### 8.3 Limites de l'analyse

- **Bornes du filtre des auteurs.** Les seuils exacts ne sont pas publiés et ont été reconstruits. La correspondance d'environ 96 % constitue une forte présomption et non une preuve formelle.
- **Absence des polygones GADM.** La cohérence spatiale n'a été évaluée qu'au moyen de signatures d'erreur et d'un cadre géographique approximatif. Une jointure spatiale reste nécessaire pour valider chaque point.
- **Calendrier éthiopien.** L'hypothèse repose sur des régularités statistiques et sur une connaissance externe du calendrier agricole. Elle doit être confirmée dans la documentation de l'enquête ESS.
- **Comparaisons entre sources.** Les rapports de la section 7.2 portent sur des cellules d'au moins 10 observations, sans pondération. Ils établissent le sens de l'écart sans en estimer l'ampleur.
- **Variables non documentées.** `prcLeft`, nul dans 99 % des cas, `season_name` et `crop_production_system` ne disposent d'aucune définition officielle. Leur utilisation reste subordonnée à une documentation externe.

#### 8.4 Décision

**Go pour la phase de Data Preparation**, sous trois conditions.

1. Les fichiers sont traités séparément. Aucune fusion n'intervient tant que la question des biais entre sources n'est pas tranchée, conformément au constat 14.
2. Les actions de préparation 3, 4, 5, 6, 8, 9, 11, 13 et 18 sont implémentées sous forme de fonctions testées dans `src/` et appliquées à une copie des données brutes.
3. Le protocole d'évaluation, fondé sur les groupes spatiaux, les ménages et les périodes, est arrêté avant l'entraînement du premier modèle.

**Recommandation relative à la cible, soumise à la validation du porteur du projet**

- Pour l'objectif de généralisation spatiale et temporelle, `yieldTonHa` dans `regional`, restreint aux niveaux infranationaux 1 et 2, offre la couverture la plus large avec 47 pays sur la période 1960–2023.
- `YieldUsingGPS_ton_ha_` dans `survey` constitue la cible la plus pertinente à l'échelle de la parcelle.
- `cropcut` sert de référence de validation pour l'Éthiopie.

Les quatre fichiers ne contiennent aucune variable climatique, pédologique ou relative aux pratiques culturales. Les prédicteurs admissibles se limitent à la culture, au lieu, à l'année et à la source. L'obtention d'une performance utile dépendra donc de l'intégration de covariables externes, climatiques et issues de la télédétection, lors du feature engineering.
