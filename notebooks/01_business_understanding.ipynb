**African Crop Yield Intelligence** : étude de la prédiction des rendements agricoles en Afrique par
apprentissage automatique, à partir du jeu de données GROW-Africa.

## Compréhension métier (Business Understanding)

**African Crop Yield Intelligence** est un projet d’étude de la prédiction des rendements agricoles en Afrique à partir du jeu de données GROW-Africa. Il vise à concevoir et à évaluer une démarche reproductible de Machine Learning, en accordant une attention particulière à la qualité des données, à la généralisation des modèles et à la fiabilité des résultats.

## 1. Contexte

L’agriculture joue un rôle important dans les moyens de subsistance, l’approvisionnement alimentaire et les économies de nombreux pays africains. La connaissance des rendements agricoles peut contribuer à l’analyse de la production, à la planification des ressources et à l’étude des risques agricoles.

Cependant, les rendements varient selon de nombreux facteurs, notamment les conditions climatiques, les caractéristiques des cultures, les pratiques agricoles et les contextes géographiques. La disponibilité et la qualité des données peuvent également varier selon les pays, les périodes et les cultures.

Dans ce contexte, la prédiction des rendements constitue un problème d’apprentissage automatique dont la résolution nécessite une analyse rigoureuse des données et une évaluation adaptée aux conditions réelles d’utilisation.

## 2. Problématique agricole

Comment construire et évaluer des modèles capables d’estimer les rendements agricoles à partir des données disponibles, tout en mesurant leur capacité à généraliser à des régions et à des périodes non vues pendant l’entraînement ?

Cette problématique implique également de déterminer les conditions dans lesquelles les prédictions sont suffisamment fiables, ainsi que les limites liées à la qualité, à la couverture et à la représentativité des données.

## 3. Objectif général

Concevoir, expérimenter et documenter une chaîne reproductible de prédiction des rendements agricoles à partir du jeu de données GROW-Africa.

Le projet vise à comparer des approches de Machine Learning selon un protocole d’évaluation explicite, à mesurer leurs performances et à examiner leur capacité de généralisation spatiale et temporelle. Il vise également à documenter les limites des modèles, à étudier les facteurs associés aux prédictions et à évaluer, lorsque les données et les méthodes le permettent, l’incertitude des estimations.

## 4. Objectifs spécifiques

- Comprendre la structure, la provenance et la signification des variables du dataset.
- Évaluer la qualité, la cohérence et la représentativité des données disponibles.
- Définir la variable cible et les informations admissibles pour la prédiction.
- Préparer les données en justifiant les décisions de traitement et de sélection des variables.
- Construire une référence de comparaison (*baseline*) avant d’évaluer des modèles plus complexes.
- Comparer les modèles avec des stratégies de validation adaptées aux objectifs de généralisation.
- Analyser les erreurs de prédiction et les limites des résultats.
- Documenter les expériences afin de permettre leur reproduction et leur vérification.

## 5. Périmètre de l’étude

L’étude repose sur le jeu de données GROW-Africa. La couverture géographique, temporelle et agricole effectivement exploitable sera établie à partir de l’inventaire des fichiers et de l’audit des données.

La variable cible envisagée est le rendement agricole, exprimé dans l’unité définie par la documentation du dataset. Sa définition exacte, sa méthode de calcul et sa disponibilité seront vérifiées avant la modélisation.

Le projet distinguera les informations disponibles au moment de la prédiction de celles qui sont produites pendant ou après la récolte. Cette distinction est nécessaire pour éviter les fuites de cible et évaluer un scénario de prédiction cohérent.

## 6. Critères de réussite

La réussite du projet sera appréciée selon plusieurs dimensions :

- **Qualité des données :** documentation des anomalies, des valeurs manquantes et des limites de couverture.
- **Rigueur expérimentale :** séparation appropriée des données, prévention des fuites d’information et comparaison équitable des modèles.
- **Performance prédictive :** évaluation à l’aide de métriques adaptées à la prédiction d’une variable numérique continue.
- **Généralisation :** mesure des performances sur des régions ou des périodes non vues, lorsque les données permettent de construire ces évaluations.
- **Reproductibilité :** traçabilité des données, des traitements, des configurations et des expériences.
- **Analyse critique :** documentation des erreurs, des incertitudes et des limites d’utilisation.

Les seuils de performance et les critères quantitatifs définitifs seront établis après l’analyse exploratoire et la définition du protocole d’évaluation.

## 8. Hypothèses et limites initiales

La faisabilité de la prédiction dépendra de la qualité et de la couverture des données, de la disponibilité des variables pertinentes et de la cohérence entre les données d’entraînement et les situations visées.

Le projet ne supposera pas qu’un modèle performant sur un échantillon donné sera nécessairement fiable dans toutes les régions, toutes les années ou pour toutes les cultures. Ces capacités devront être évaluées empiriquement.

Les résultats seront interprétés dans les limites du dataset, du protocole expérimental et des conditions de validation retenues.
