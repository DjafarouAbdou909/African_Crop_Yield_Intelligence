## Compréhension métier (Business Understanding)
**African Crop Yield Intelligence** : étude de la prédiction des rendements agricoles en Afrique par
apprentissage automatique, à partir du jeu de données GROW-Africa.

## Contexte
L'agriculture occupe une place centrale dans l'économie et la sécurité alimentaire de nombreux pays africains.
Anticiper les rendements aide à la planification (stockage, approvisionnement, logistique) et à l'analyse des
risques. Or la production agricole est exposée à une forte variabilité (climat, sols, pratiques, accès aux intrants),
et les données disponibles sont souvent hétérogènes d'un pays ou d'une culture à l'autre.

## Problématique agricole
Comment estimer des rendements de façon fiable lorsque les observations sont inégalement réparties dans
l'espace, dans le temps et entre cultures, et comment savoir dans quelles conditions l'estimation reste digne de
confiance ?

## Objectif général
Concevoir, évaluer et documenter une chaîne reproductible de prédiction des rendements agricoles, en
démontrant par des expériences contrôlées dans quelle mesure les modèles de Machine Learning généralisent
à des régions et à des périodes non vues, avec explicabilité et estimation de l'incertitude.

## Public cible et utilisateurs potentiels
| Public | Usage attendu |
|---|---|
| Auteur du projet (étudiant en IA et Big Data) | Démontrer la maîtrise d'un cycle complet de Machine Learning Engineering et de recherche appliquée. |
